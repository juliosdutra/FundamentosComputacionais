# 🎨 Introdução à Visualização de Dados com Matplotlib

A biblioteca **Matplotlib** é uma das ferramentas mais populares em Python para criação de gráficos. Ela permite representar dados de forma visual, o que facilita a análise, a comunicação e a compreensão de informações complexas.

## 🧭 Por que usar gráficos?

- 📈 Compreender **tendências e padrões**
- 🔍 Investigar **relações entre variáveis**
- 🧪 Visualizar **resultados de experimentos ou simulações**
- 🧠 Apoiar **decisões técnicas** e **comunicação científica**

## 🧩 Anatomia de um gráfico

Um gráfico em Matplotlib geralmente envolve:

- **Figura (`Figure`)**: a área total de desenho (janela do gráfico)
- **Eixos (`Axis`)**: região onde os dados são desenhados (com eixos x e y)
- **Curvas e marcadores (`Lines`)**: os dados propriamente ditos
- **Rótulos (`Labels`)**: nomes dos eixos, título e legendas
- **Grade (`Grid`)**: linhas de referência opcionais
- **Barra de cores (`Colorbar`)**: útil em mapas de calor

# **📈 Gráfico de linhas - `plt.plot()`**

O `plt.plot()` é usado para visualizar dados contínuos, como variações ao longo do tempo ou resultados de simulações.

Este exemplo mostra como sobrepor duas curvas simples, aplicando boas práticas de visualização: uso de `label`, `grid`, título, rótulos de eixos e `tight_layout()` para melhor espaçamento.



````python
# Importa as bibliotecas
import numpy as np
import matplotlib.pyplot as plt

# Cria a variável independente x
var_x = np.array([0, 1, 2, 3])

# Cria as variáveis dependentes y
y1 = [0, 1, 2, 3]              # Função linear
y2 = [x**2 + 1 for x in var_x] # Função quadrática

# Cria a figura com largura 6 e altura 3
plt.figure(figsize=(6, 3))

# Plota a curva linear
# 'b-' -> linha azul contínua
plt.plot(var_x, y1, 'b-', label='Linear')

# Plota a curva quadrática
# 'r--' -> linha vermelha tracejada
plt.plot(var_x, y2, 'r--', label='Quadrática')

# Define os nomes dos eixos
plt.xlabel('Variável independente')
plt.ylabel('Valor de resposta')

# Define o título do gráfico
plt.title('Exemplo de Gráfico')

# Exibe a legenda das curvas
plt.legend()

# Exibe grade no gráfico
plt.grid(True)

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

# **Múltiplos gráficos - `plt.subplot(m,n,p)`**

O comando `plt.subplot(m, n, p)` divide a figura em uma grade de `m` linhas e `n` colunas, e ativa a posição `p` para desenhar o gráfico. Isso é útil para comparar diferentes visualizações lado a lado.

````python
# Cria o vetor de entrada
x = np.linspace(0, 2 * np.pi, 100)

# Calcula as funções
y1 = np.sin(x)               # Função seno
y2 = np.cos(x)               # Função cosseno
y3 = np.sin(x) + np.cos(x)   # Soma das funções

# Cria a figura
plt.figure(figsize=(6, 6))

# ==========================================================
# GRÁFICO 1
# ==========================================================

# Cria um subplot:
# (3 linhas, 1 coluna, posição 1)
plt.subplot(3, 1, 1)

# Plota a função seno
# 'b' -> azul
plt.plot(x, y1, 'b')

# Define título e eixo y
plt.title('Função seno')
plt.ylabel('sin(x)')


# ==========================================================
# GRÁFICO 2
# ==========================================================

# Cria o segundo subplot
plt.subplot(3, 1, 2)

# Plota a função cosseno
# 'r' -> vermelho
plt.plot(x, y2, 'r')

# Define título e eixo y
plt.title('Função cosseno')
plt.ylabel('cos(x)')


# ==========================================================
# GRÁFICO 3
# ==========================================================

# Cria o terceiro subplot
plt.subplot(3, 1, 3)

# Plota a soma das funções
# 'g' -> verde
plt.plot(x, y3, 'g')

# Define título e nomes dos eixos
plt.title('Seno + Cosseno')
plt.xlabel('x')
plt.ylabel('sin(x) + cos(x)')


# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe a figura
plt.show()
````

**Gráfico de Padrão de Borboleta + Coração 🦋❤️**

````python
# Importa as bibliotecas
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# PADRÃO DE BORBOLETA
# ==========================================================

# Cria o vetor de parâmetros
t1 = np.linspace(0, 2 * np.pi, 300)

# Define as coordenadas x e y da curva
x1 = np.sin(t1) * (np.exp(np.cos(t1)) - 2 * np.cos(4 * t1))
y1 = np.cos(t1) * (np.exp(np.cos(t1)) - 2 * np.cos(4 * t1))


# ==========================================================
# PADRÃO DE CORAÇÃO
# ==========================================================

# Cria o vetor de parâmetros
t2 = np.linspace(0, 2 * np.pi, 500)

# Define as coordenadas x e y da curva
x2 = 16 * np.sin(t2)**3
y2 = (
    13 * np.cos(t2)
    - 5 * np.cos(2*t2)
    - 2 * np.cos(3*t2)
    - np.cos(4*t2)
)


# ==========================================================
# CRIAÇÃO DA FIGURA
# ==========================================================

# Cria uma figura com:
# - 1 linha
# - 2 colunas
# - tamanho 6x3
fig, axs = plt.subplots(1, 2, figsize=(6, 3))


# ==========================================================
# SUBPLOT 1 - BORBOLETA
# ==========================================================

# Plota a primeira metade da borboleta
axs[0].plot(x1, y1, color='purple', linewidth=2)

# Plota a parte espelhada da borboleta
axs[0].plot(-x1, y1, color='orange', linewidth=2)

# Define o título do gráfico
axs[0].set_title("Padrão de Borboleta", fontsize=14)

# Mantém a mesma escala nos eixos
axs[0].axis("equal")

# Remove os eixos da figura
axs[0].axis("off")


# ==========================================================
# SUBPLOT 2 - CORAÇÃO
# ==========================================================

# Plota o padrão de coração
axs[1].plot(x2, y2, color='red', linewidth=4)

# Define o título do gráfico
axs[1].set_title("Padrão de Coração", fontsize=14)

# Mantém a mesma escala nos eixos
axs[1].axis("equal")

# Remove os eixos da figura
axs[1].axis("off")


# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe a figura
plt.show()
````

# **Gráfico em 2D/3D**

## **Como usar `contourf()`, `plot_surface` e `contour3D`?**

> 💡 Explicação Rápida:
* `contourf` → visualização 2D com níveis de cor preenchidos.
* `plot_surface` → representação 3D da mesma função.
* `contour3D` → contorno em 3D com elevações e linhas de nível sobre uma malha 3D.
* `cm.viridis_r` → colormap invertido para melhor contraste.



**Exemplo simples**

````python
# Cria uma matriz 8x8 com valores aleatórios
# entre 0 e 1
Z = np.random.uniform(0, 1, (8, 8))

# Cria um gráfico de contorno preenchido
# usando o mapa de cores 'viridis_r'
plt.contourf(Z, cmap='viridis_r')

# Exibe a barra de cores
# associada aos valores da matriz
plt.colorbar()

# Exibe o gráfico
plt.show()
````

**Exemplo com função $Z=f(X,Y)$**

````python
# Importa ferramentas para gráficos 3D
from mpl_toolkits.mplot3d import Axes3D

# ==========================================================
# CRIAÇÃO DA MALHA 2D
# ==========================================================

# Define o número de pontos em x e y
nx, ny = 100, 100

# Cria os vetores de coordenadas
x = np.linspace(0, 0.909, nx)
y = np.linspace(0, 0.909, ny)

# Cria a grade bidimensional
# combinando x e y
X, Y = np.meshgrid(x, y)


# ==========================================================
# DEFINIÇÃO DA SUPERFÍCIE
# ==========================================================

# Define um paraboloide deslocado
Z = (X - 0.2)**2 + (Y - 0.2)**2


# ==========================================================
# CRIAÇÃO DA FIGURA
# ==========================================================

# Cria a figura com tamanho 12x5
fig = plt.figure(figsize=(12, 5))


# ==========================================================
# SUBPLOT 1 - MAPA DE CONTORNO (2D)
# ==========================================================

# Cria o primeiro subplot
ax1 = fig.add_subplot(1, 2, 1)

# Cria o gráfico de contorno preenchido
cont = ax1.contourf(x, y, Z, cmap='viridis_r')

# Adiciona barra de cores
cbar1 = fig.colorbar(cont, ax=ax1)

# Define o nome da barra de cores
cbar1.set_label('z = f(x, y)')

# Define nomes dos eixos
ax1.set_xlabel('X')
ax1.set_ylabel('Y')

# Define o título
ax1.set_title('Mapa de Contorno')


# ==========================================================
# SUBPLOT 2 - SUPERFÍCIE 3D
# ==========================================================

# Cria o segundo subplot em 3D
ax2 = fig.add_subplot(1, 2, 2, projection='3d')

# Cria o gráfico de superfície
surf = ax2.plot_surface(X, Y, Z, cmap='viridis_r')

# Adiciona barra de cores
cbar2 = fig.colorbar(
    surf,
    ax=ax2,
    shrink=0.5,
    aspect=10,
    pad=0.1
)

# Define o nome da barra de cores
cbar2.set_label('z = f(x, y)')

# Define nomes dos eixos
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('z = f(x, y)')

# Define o título
ax2.set_title('Paraboloide')


# ==========================================================
# FINALIZAÇÃO
# ==========================================================

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe a figura
plt.show()
````

**Gráfico de contorno em 3D**

Este código cria um contorno 3D para a função $sin(\sqrt{x^2+y^2})$.

````python
# Importa as bibliotecas
import numpy as np
import matplotlib.pyplot as plt

# Importa ferramentas para gráficos 3D
from mpl_toolkits.mplot3d import Axes3D


# ==========================================================
# CRIAÇÃO DA MALHA 2D
# ==========================================================

# Cria vetores no intervalo de -5 até 5
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

# Cria a grade bidimensional
X, Y = np.meshgrid(x, y)


# ==========================================================
# DEFINIÇÃO DA FUNÇÃO
# ==========================================================

# Define a função matemática
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# Calcula os valores da superfície
Z = f(X, Y)


# ==========================================================
# CRIAÇÃO DA FIGURA 3D
# ==========================================================

# Cria a figura
fig = plt.figure(figsize=(15, 6))

# Cria um subplot 3D
ax = fig.add_subplot(111, projection='3d')


# ==========================================================
# GRÁFICO DE CONTORNO 3D
# ==========================================================

# Cria o gráfico de contorno 3D
# 50 -> número de níveis de contorno
contorno = ax.contour3D(X, Y, Z, 50, cmap='viridis')


# ==========================================================
# CUSTOMIZAÇÃO
# ==========================================================

# Define os nomes dos eixos
ax.set_xlabel('Eixo X')
ax.set_ylabel('Eixo Y')
ax.set_zlabel('Eixo Z')

# Adiciona barra de cores
fig.colorbar(contorno, ax=ax, label='Valores de Z')


# ==========================================================
# EXIBIÇÃO
# ==========================================================

# Exibe o gráfico
plt.show()
````

# **Salvar gráficos**

Para salvar um gráfico em Python com matplotlib, você pode usar a função:

```python
plt.savefig('nome_do_arquivo.png', dpi=300, bbox_inches='tight')


> Explicação dos argumentos:
* 'nome_do_arquivo.png': nome e extensão do arquivo (pode ser .png, .jpg, .pdf, .svg, etc.)
* dpi=300: define a resolução (300 é ideal para impressão)
* bbox_inches='tight': recorta espaços em branco ao redor da imagem

````python
# Importa bibliotecas
from google.colab import files
import os
import numpy as np
import matplotlib.pyplot as plt


# ==========================================================
# CRIAÇÃO DOS DADOS
# ==========================================================

# Cria os valores de x
x = np.linspace(0, 10, 100)

# Calcula sen(x)
y = np.sin(x)


# ==========================================================
# CRIAÇÃO DO GRÁFICO
# ==========================================================

# Cria o gráfico
# 'C1' -> segunda cor padrão do Matplotlib
plt.plot(x, y, 'C1')

# Define título e nomes dos eixos
plt.title('Gráfico de seno')
plt.xlabel('x')
plt.ylabel('sen(x)')


# ==========================================================
# SALVANDO A FIGURA
# ==========================================================

# Salva o gráfico em arquivo PNG
#
# dpi=300: resolução da imagem#
# bbox_inches='tight': remove espaços extras nas bordas

plt.savefig(
    'grafico_seno.png',
    dpi=300,
    bbox_inches='tight'
)

# Exibe o gráfico
plt.show()


# ==========================================================
# DOWNLOAD DO ARQUIVO NO GOOGLE COLAB
# ==========================================================

# Verifica se o arquivo existe
if os.path.exists('grafico_seno.png'):

    print("O arquivo foi criado com sucesso.")

    # Faz download do arquivo
    files.download('grafico_seno.png')

else:

    print("O arquivo não foi encontrado.")
````

## **🔸 Gráfico de Dispersão - `plt.scatter()`**

O `plt.scatter()` é útil para visualizar a relação entre duas variáveis, especialmente em dados experimentais com ruído.

É ideal para detectar correlações, padrões ou agrupamentos nos dados.

````python
# Cria 50 valores aleatórios para x
x = np.random.rand(50)

# Cria valores de y com relação linear
# e adiciona ruído aleatório
y = 2 * x + np.random.normal(0, 0.1, 50)

# Cria um gráfico de dispersão
# usando pontos na cor laranja
plt.scatter(x, y, color='darkorange')

# Define os nomes dos eixos
plt.xlabel('x')
plt.ylabel('y')

# Define o título do gráfico
plt.title('Exemplo de Gráfico de Dispersão')

# Exibe grade no gráfico
plt.grid(True)

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

## **🟦 Área Sombreada - `plt.fill_between()`**


O `plt.fill_between()` preenche a área entre duas curvas ou entre uma curva e um valor fixo.

É ideal para mostrar:
- Incertezas experimentais
- Intervalos de confiança
- Faixas de operação segura

````python
# Cria os valores do eixo x
x = np.linspace(0, 10, 100)

# Define a curva central
y_central = np.sin(x)

# Define os limites superior e inferior
# da faixa de incerteza
y_sup = y_central + 0.3
y_inf = y_central - 0.3

# Cria a curva principal
plt.plot(
    x,
    y_central,
    label='Média',
    color='blue'
)

# Cria uma área sombreada entre
# os limites inferior e superior
#
# alpha: controla a transparência
plt.fill_between(
    x,
    y_inf,
    y_sup,
    color='blue',
    alpha=0.2,
    label='Incerteza'
)

# Define os nomes dos eixos
plt.xlabel('x')
plt.ylabel('y')

# Define o título do gráfico
plt.title('Exemplo de Área Sombreada')

# Exibe a legenda
plt.legend()

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

## **🔥Imagem ou mapa de calor - `plt.imshow()`**
* Ideal para matrizes, campos, resultados de simulações numéricas (ex: CFD, transferência de calor).
* Muito usado para visualizar mapas 2D (temperatura, concentração).


````python
# ==========================================================
# CRIAÇÃO DA MALHA ESPACIAL E TEMPORAL
# ==========================================================

# Define as posições espaciais
x = np.linspace(0, 1, 20)

# Define os instantes de tempo
t = np.linspace(0, 10, 50)

# Cria a grade bidimensional
# combinando espaço e tempo
T, X = np.meshgrid(t, x)


# ==========================================================
# SIMULAÇÃO DA TEMPERATURA
# ==========================================================

# Define um campo de temperatura
# que decai com o tempo
temperatura = (
    np.sin(np.pi * X)
    * np.exp(-0.3 * T)
)


# ==========================================================
# VISUALIZAÇÃO COM IMAGEM
# ==========================================================

# Cria a figura
plt.figure(figsize=(8, 5))

# Cria o mapa de calor
#
# aspect='auto': ajusta automaticamente as proporções
#
# extent: define os limites dos eixos
#
# cmap: define o mapa de cores
im = plt.imshow(
    temperatura,
    aspect='auto',
    extent=[
        x.min(),
        x.max(),
        t.min(),
        t.max()
    ],
    cmap='inferno'
)

# Adiciona barra de cores
plt.colorbar(label='Temperatura')

# Define os nomes dos eixos
plt.ylabel('Posição (x)')
plt.xlabel('Tempo (t)')

# Define o título
plt.title('Mapa de Calor: Temperatura T(t, x)')

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

###**🔁 Código da animação**

````python
# ==========================================================
# IMPORTAÇÃO DAS BIBLIOTECAS
# ==========================================================

# Biblioteca numérica
import numpy as np

# Biblioteca de gráficos
import matplotlib.pyplot as plt

# Biblioteca para animações
import matplotlib.animation as animation

# Permite exibir animações em notebooks
from IPython.display import HTML


# ==========================================================
# CRIAÇÃO DA MALHA ESPACIAL E TEMPORAL
# ==========================================================

# Define as posições espaciais
x = np.linspace(0, 1, 50)

# Define os instantes de tempo
t = np.linspace(0, 10, 100)

# Cria a grade bidimensional
T, X = np.meshgrid(t, x)


# ==========================================================
# SIMULAÇÃO DA TEMPERATURA
# ==========================================================

# Define um campo de temperatura
# que decai com o tempo
temperatura = (
    np.sin(np.pi * X)
    * np.exp(-0.3 * T)
)


# ==========================================================
# CONFIGURAÇÃO INICIAL DA FIGURA
# ==========================================================

# Define um pequeno intervalo inicial
delta_t = (t[1] - t[0]) if len(t) > 1 else 0.1

# Cria a figura e os eixos
fig, ax = plt.subplots(figsize=(8, 4))

# Cria a imagem inicial
img = ax.imshow(
    temperatura[:, 0:2],

    # Ajusta automaticamente a proporção
    aspect='auto',

    # Define limites dos eixos
    extent=[
        t[0],
        t[0] + delta_t,
        x.min(),
        x.max()
    ],

    # Define o mapa de cores
    cmap='inferno',

    # Define limites da escala de cores
    vmin=0,
    vmax=1
)


# ==========================================================
# CUSTOMIZAÇÃO DA FIGURA
# ==========================================================

# Define nomes dos eixos
ax.set_xlabel('Tempo (t)')
ax.set_ylabel('Posição (x)')

# Define o título
ax.set_title('Evolução da Temperatura T(t, x)')

# Adiciona barra de cores
cbar = fig.colorbar(img, ax=ax)

# Define o nome da barra de cores
cbar.set_label('Temperatura')


# ==========================================================
# FUNÇÃO DE ATUALIZAÇÃO DA ANIMAÇÃO
# ==========================================================

# Atualiza a imagem a cada frame
def update(frame):

    # Define a largura inicial
    if frame < 2:
        width = delta_t

    # Atualiza a largura ao longo do tempo
    else:
        width = t[frame] - t[0]

    # Atualiza os dados exibidos
    img.set_data(temperatura[:, 0:frame+1])

    # Atualiza os limites da imagem
    img.set_extent([
        t[0],
        t[0] + width,
        x.min(),
        x.max()
    ])

    return [img]


# ==========================================================
# CRIAÇÃO DA ANIMAÇÃO
# ==========================================================

# Cria a animação
#
# frames:
# número de quadros
#
# interval:
# tempo entre quadros em ms
ani = animation.FuncAnimation(
    fig,
    update,
    frames=len(t),
    interval=100,
    blit=True
)

# Evita exibir figura estática duplicada
plt.close(fig)


# ==========================================================
# EXIBIÇÃO DA ANIMAÇÃO
# ==========================================================

# Exibe a animação no notebook
HTML(ani.to_jshtml())
````

#### ✅ 2. Gráfico de dispersão com `sns.scatterplot()`

* ideal para análise de correlação e agrupamentos



````python
# Importa as bibliotecas
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np


# ==========================================================
# CRIAÇÃO DOS DADOS
# ==========================================================

# Define a semente aleatória
# para reproduzir os mesmos resultados
np.random.seed(42)

# Número de amostras
n = 50

# Cria um DataFrame com dados simulados
df = pd.DataFrame({

    # Vazão com distribuição normal
    'vazao': np.random.normal(100, 10, n),

    # Concentração com distribuição normal
    'concentracao': np.random.normal(0.5, 0.1, n),

    # Grupo categórico aleatório
    'grupo': np.random.choice(['A', 'B'], n)
})


# ==========================================================
# GRÁFICO DE DISPERSÃO
# ==========================================================

# Cria um gráfico de dispersão
#
# hue:
# define cores diferentes por grupo
#
# style:
# define marcadores diferentes por grupo
sns.scatterplot(
    data=df,
    x='vazao',
    y='concentracao',
    hue='grupo',
    style='grupo'
)

# Define o título do gráfico
plt.title('Relação Vazão x Concentração')

# Define os nomes dos eixos
plt.xlabel('Vazão (kg/h)')
plt.ylabel('Concentração (fração mássica)')

# Exibe grade
plt.grid(True)

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

#### ✅ 3. Gráfico de mapa de calor com `sns.heatmap()`

* é uma ferramenta utilizada para visualizar valores organizados em uma matriz, permitindo identificar rapidamente padrões, correlações e regiões críticas.


````python
# Importa as bibliotecas
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


# ==========================================================
# CRIAÇÃO DOS DADOS
# ==========================================================

# Cria um dicionário com valores simulados
dados = {

    # Cada chave representa o número
    # de camadas ocultas
    8:  [1.249080, 0.812037, 1.702230],
    12: [2.401429, 0.811989, 1.916145],
    16: [1.963988, 0.616167, 40.000000],
    20: [1.697317, 2.232352, 2.439820],
}

# Define os índices da tabela
indice = [3, 4, 5]

# Cria o DataFrame
df = pd.DataFrame(dados, index=indice)

# Define nomes dos índices e colunas
df.index.name = "Nº de neurônios"
df.columns.name = "Camadas ocultas"


# ==========================================================
# HEATMAP
# ==========================================================

# Cria a figura
plt.figure(figsize=(6, 4))

# Cria o mapa de calor
#
# annot=True:
# exibe os valores nas células
#
# fmt=".2f":
# mostra os valores com 2 casas decimais
#
# cmap:
# define o mapa de cores
#
# vmax=5:
# limita o valor máximo da escala de cores
sns.heatmap(
    df,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmax=5,

    # Define o nome da barra de cores
    cbar_kws={
        'label': 'Erro Quadrático Médio'
    }
)

# Define o título
plt.title("Erro quadrático médio dos modelos (valores simulados)")

# Define nomes dos eixos
plt.xlabel("Camadas ocultas")
plt.ylabel("Nº de neurônios")

# Ajusta automaticamente os espaçamentos
plt.tight_layout()

# Exibe o gráfico
plt.show()
````

````python
# Passo 2: Exemplo de Gráfico Interativo

import numpy as np
import pyvista as pv
from IPython.display import Image

# Gerar dados
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
x, y = np.meshgrid(x, y)

# Superfície
z = np.sin(3 * np.sqrt(x**2 + y**2))/ (1 + x**2 + y**2)

# Criar grade estruturada
grid = pv.StructuredGrid(x, y, np.zeros_like(z))  # Altura inicial plana
grid["altura"] = z.ravel(order="F")  # Adiciona como escalar

# Deformar a superfície com base no escalar
surface = grid.warp_by_scalar("altura", factor=1.0)

# Criar plotter off-screen
plotter = pv.Plotter(off_screen=True)
plotter.add_mesh(surface, cmap='inferno', show_edges=False)
plotter.add_axes()
plotter.add_text("Superfície: z = f(x, y)", font_size=12)
plotter.set_background("white")
plotter.show(screenshot='superficie.png')

# Exibir imagem
Image('superficie.png')
````
